# 🔹 Eigenvalues & Eigenvectors

- An **eigenvector** of a matrix is a vector whose direction does not change after transformation by the matrix.
- It may only stretch or shrink.

- The **eigenvalue** tells us how much the eigenvector is scaled.

---

For a square matrix A and vector v:

**A v = λ v**

Where:
- A = transformation matrix
- v = eigenvector
- λ = eigenvalue

---

**Interpretation**

After transformation:
- Direction stays same
- Only magnitude changes

Example:

If: λ = 3, then Vector becomes **3 times larger**.


## 📘 Real-World Example: Feature Transformation Matrix


```python
# Eigenvalues & Eigenvectors

import numpy as np

# Define matrix
A = np.array([
    [2, 1],
    [1, 2]
])

# -----------------------------
# Step 1: Eigenvalues & Eigenvectors
# -----------------------------
eigenvalues, eigenvectors = np.linalg.eig(A)

# -----------------------------
# Step 2: Verify Av = λv
# -----------------------------
v = eigenvectors[:,0]
lambda_val = eigenvalues[0]

Av = A @ v
lambda_v = lambda_val * v

# -----------------------------
# Output
# -----------------------------
print("Matrix A:\n", A)

print("\nEigenvalues:")
print(eigenvalues)

print("\nEigenvectors:")
print(eigenvectors)

print("\nVerification (Av):", Av)
print("λv:", lambda_v)
```
